# Netflix Data Cleaning Project

## Step 1: Import Libraries

```
import pandas as pd
import numpy as np
```

## Step 2: Load Dataset

```
df = pd.read_csv('netflix_titles.csv')
df.head()
```

## Step 3: Dataset Overview

```
df.shape

df.info()

df.describe(include='all')
```

## Step 4: Check Missing Values

```
df.isnull().sum()
```

## Step 5: Check Duplicate Rows

```
df.duplicated().sum()
```

## Step 6: Remove Unnecessary Columns

```
df.drop(columns=['show_id','description'], inplace=True)
```

## Step 7: Rename Columns

```
df.rename(columns={
    'type':'content_type',
    'cast':'cast_members',
    'listed_in':'genres',
    'date_added':'date_added_to_netflix'
}, inplace=True)
```

## Step 8: Remove Duplicate Records

```
df.drop_duplicates(inplace=True)
```

## Step 9: Handle Missing Values

```
df['director']=df['director'].fillna('Unknown')
df['cast_members']=df['cast_members'].fillna('Unknown')
df['country']=df['country'].fillna('Unknown')
df['rating']=df['rating'].fillna('Unknown')
df['duration']=df['duration'].fillna('Unknown')
```

## Step 10: Convert Data Types

```
df['date_added_to_netflix']=pd.to_datetime(df['date_added_to_netflix'], errors='coerce')
df['release_year']=df['release_year'].astype('Int64')
```

## Step 11: Remove Extra Spaces

```
text_cols=df.select_dtypes(include='object').columns
for col in text_cols:
    df[col]=df[col].str.strip()
```

## Step 12: Verify Cleaned Data

```
df.info()
df.head()
df.isnull().sum()
```

## Step 13: Export Clean Dataset

```
df.to_csv('netflix_cleaned.csv', index=False)
```
